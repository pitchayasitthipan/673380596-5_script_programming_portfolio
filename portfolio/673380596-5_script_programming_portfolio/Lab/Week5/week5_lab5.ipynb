## Lab 5.1

```
# lab5_1.py
# Lab 5.1: Simple Contact Book Manager (Dictionary)
# ชื่อ: นส.พิชยา สิทธิพันธ์  รหัสนักศึกษา: 673380596-5  sec.1

contacts = {}

while True:
    print("\n--- Contact Book ---")
    print("1. Add/Update Contact")
    print("2. View Contact Details")
    print("3. List All Contacts")
    print("4. Delete Contact")
    print("5. Exit")
    choice = input("Choose (1-5): ")

    if choice == "1":
        # ใช้ .title() ให้ชื่อเป็นแบบเดียวกันเสมอ (case-insensitive)
        name = input("Enter name: ").strip().title()
        if name == "":
            print("Name cannot be empty")
        else:
            phone = input("Enter phone: ")
            email = input("Enter email: ")
            if name in contacts:
                print("Updated", name)
            else:
                print("Added", name)
            contacts[name] = {"phone": phone, "email": email}

    elif choice == "2":
        name = input("Enter name to view: ").strip().title()
        details = contacts.get(name)
        if details is None:
            print("Contact not found.")
        else:
            print(name, "->", details)

    elif choice == "3":
        if len(contacts) == 0:
            print("No contacts yet")
        else:
            for name, details in contacts.items():
                print(name, ":", details)

    elif choice == "4":
        name = input("Enter name to delete: ").strip().title()
        if name in contacts:
            contacts.pop(name)
            print("Deleted", name)
        else:
            print("Contact not found.")

    elif choice == "5":
        print("Bye!")
        break

    else:
        print("Please choose 1-5")
```

## Lab 5.2

```
# lab5_2.py
# Lab 5.2: Course Enrollment & Collaboration (Set)
# ชื่อ: นส.พิชยา สิทธิพันธ์  รหัสนักศึกษา: 673380596-5  sec.1

python_students = {"S001", "S003", "S005", "S007", "S009"}
java_students = {"S002", "S003", "S006", "S007", "S010"}

# นักเรียนที่เรียนทั้งสองวิชา
both = python_students.intersection(java_students)
print("Students in both courses:", sorted(both))

# นักเรียนทั้งหมด (ไม่ซ้ำ)
all_students = python_students.union(java_students)
print("All unique students:", sorted(all_students))

# เรียน Python อย่างเดียว
only_python = python_students.difference(java_students)
print("Only in Python:", sorted(only_python))

# เรียนแค่วิชาเดียว (ไม่ได้เรียนทั้งคู่)
only_one = python_students.symmetric_difference(java_students)
print("Only in one course:", sorted(only_one))

# ลบค่าซ้ำออกจาก list
all_students_list = ["S001", "S003", "S005", "S001", "S002"]
unique_list = list(set(all_students_list))
print("Original list:", all_students_list)
print("Unique list:", unique_list)
```

# Activity 5.1: Real-World Applications of Dictionaries and Sets

**ชื่อ:** นส.พิชยา สิทธิพันธ์  **รหัสนักศึกษา:** 673380596-5  **sec.1**

**1. ใช้ dictionary ในเกมตรงไหน**
เก็บสถานะตัวละครหรือของในกระเป๋า เช่น `{"hp": 100, "mp": 50, "level": 5}` หรือ `{"potion": 3, "sword": 1}`
เพราะเรียกดูด้วยชื่อ (key) ได้ตรงๆ เช่น `stats["hp"]` ไม่ต้องจำว่าค่าอยู่ตำแหน่งที่เท่าไร และแก้ค่าได้ง่ายเมื่อเกมดำเนินไป

**2. dictionary กับข้อมูลสินค้าในร้านออนไลน์**
ใช้รหัสสินค้าเป็น key และเก็บรายละเอียดเป็น dictionary ซ้อนข้างใน เช่น
`{"P001": {"name": "Mouse", "price": 350, "stock": 20}}`
เพราะค้นหาสินค้าจากรหัสได้เร็ว (`products["P001"]`) และแต่ละสินค้าเก็บข้อมูลได้หลายอย่าง

**3. set ใช้จัดการสิทธิ์ผู้ใช้หรือแท็กตรงไหน**
เก็บบทบาท (role) ของผู้ใช้ เช่น `{"admin", "editor"}` หรือแฮชแท็กของโพสต์
เพราะ set ไม่เก็บค่าซ้ำ ถึงเพิ่ม role หรือแท็กเดิมซ้ำก็ไม่เกิดค่าซ้ำ และเช็กว่ามีสิทธิ์นั้นอยู่ไหมได้ด้วย `in`

**4. ตรวจว่ามีของอยู่ในชุดข้อมูลใหญ่ไหม โดยไม่สนลำดับ**
เช่น รายการ IP ที่ถูกบล็อก หรือจำนวนผู้เข้าชมเว็บที่ไม่ซ้ำกัน
ใช้ set เพราะเช็ก `ip in blocked_ips` ได้เร็วแม้ข้อมูลเยอะ และไม่ต้องสนใจลำดับ ส่วนค่าซ้ำถูกตัดทิ้งให้อัตโนมัติ

**สรุป:** dictionary เหมาะกับข้อมูลที่เรียกด้วยชื่อ/รหัส (key-value) ส่วน set เหมาะกับข้อมูลที่ต้องไม่ซ้ำและต้องเช็กว่ามีอยู่หรือไม่

```
# activity5_2.py
# Activity 5.2: Word Frequency Counter (Dictionary)
# ชื่อ: นส.พิชยา สิทธิพันธ์  รหัสนักศึกษา: 673380596-5  sec.1

sentence = input("Enter a sentence: ")
sentence = sentence.lower()
words = sentence.split()

word_counts = {}

for word in words:
    # ตัดเครื่องหมายวรรคตอนที่ติดหน้า/หลังคำออก (Challenge)
    word = word.strip('.,!?"\'')
    if word == "":
        continue
    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1

print(word_counts)
```
